# 5) Split the data in training and testing sets 

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# 6) Fit the decision tree model with various parameters 

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt1 = DecisionTreeClassifier(max_depth=3, random_state=42)
dt1.fit(x_train, y_train)
```

```python
dt2 = DecisionTreeClassifier(max_depth=5, min_samples_split=5, random_state=42)
dt2.fit(x_train, y_train)
```

```python
dt_final = DecisionTreeClassifier(
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42
)
dt_final.fit(x_train, y_train)
```

# 7) Create a function to display precision score, recall score, accuracy, 
# classification report, confusion matrix, F1 Score.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import classification_report, confusion_matrix
```

```python
def evaluate_model(y_test, y_pred):
    print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall: {recall_score(y_test, y_pred):.4f}")
    print(f"F1 Score: {f1_score(y_test, y_pred):.4f}")
    
    print("\nClassification Report:\n")
    print(classification_report(y_test, y_pred))
    
    print("\nConfusion Matrix:\n")
    print(confusion_matrix(y_test, y_pred))
```

```python
# Predictions
y_pred = dt_final.predict(x_test)

# Evaluation
evaluate_model(y_test, y_pred)
```
